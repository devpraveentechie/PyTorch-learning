```python
#fixed input 
x=torch.tensor(2.0)
#leanable number
y=torch.tensor(3.0, requires_grad=True)
#predction logic
y_pred = x*y
# set some correct value as true value of y
y_true=torch.tensor(10.0)
#loss calculation
loss=(y_pred-y_true)**2

print("Before update")
print("prediction", y_pred.item())
print("loss", loss.item())
# move backward to check correct value 
loss.backward()
print("gradient", y.grad.item())
lr =0.1
# update weight
with torch.no_grad(): # donot track the y
    y -= lr*y.grad
# reset the value of 
y.grad.zero_()

print("after update")
print("y value", y.item())
```

```python
x=torch.tensor([1.,2.,3.,4.]) # fixed numbers
y_true = torch.tensor([2.,4.,6.,8.]) # correct numbers
# learning number
w= torch.tensor(0.0, requires_grad=True)
lr =0.1
epochs=10

for epoch in range(epochs):
    # step 1 prediction
    y_pred =w * x 
    #step 2 loss calcution
    loss = ((y_pred-y_true)**2).mean()

    # loss backward check
    loss.backward()
    with torch.no_grad():
        w -= lr*w.grad

    print(f"Epoch {epoch+1}: w={w.item():.4f}, loss={loss.item():.4f}")


```

```python
import torch
import torch.nn as nn
import torch.optim as optim

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.Linear = nn.Linear(1,1)

    def forward(self, x):
        return self.Linear(x)

x= torch.tensor([[1.],[2.], [3.], [4.]])
y = torch.tensor([[2.],[4.], [6.], [8.]])

model = MyModel()

loss_fn = nn.MSELoss()
# there are two kind of optimizer 1.SGD and 2. Adam example 
#x = torch.optimizer.SGD(model.parameters, 0.01) - if model require SGD optimizer, 
# and you want replace optimizer with Adam then you have pass the learning rate 10th of initial learning rate
#y= torch.optimizer.Adam(model.parameters, 0.001)
optimizer = optim.Adam(model.parameters(), lr=0.1) 

for epoch in range(10):
    optimizer.zero_grad()
    y_pred = model(x)
    loss = loss_fn(y_pred, y)
    loss.backward()
    optimizer.step()
    print(f"Epoch {epoch+1}, Loss={loss.item():.4f}")

    # save model
torch.save(model.state_dict(), "linear_model.pth")
```
